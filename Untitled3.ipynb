```python
nome = input("Digite o nome do produto: ")
peso = float(input("Digite o peso em kg: "))
valor = float(input("Digite o valor por kg: "))

valorTotal = peso * valor
desconto = valorTotal * 0.05

if(valorTotal > 100):
  valorFinal = valorTotal - desconto
else:
  valorFinal = valorTotal

print("Nome do produto", nome)
print("Valor Bruto", valorTotal)
print("Valor do desconto", desconto)
print("Valor final com desconto", valorFinal)
```

```python
nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))

media = round((nota1 + nota2 + nota3) / 3, 2)
print(media)

if(media >= 7):
  print("aprovado")
elif(media >= 5 and media <= 6.9):
  print("recuperação")
else:
  print("reprovado")
```

```python
def converteParaFahreheit(c):
  return (c * 9/5) + 32

print(converteParaFahreheit(36))
```
