```python
from fractions import Fraction
f1 = Fraction(input("Digite a primeira fração: "))
f2 = Fraction(input("Digite a segunda fração: "))
print("Soma:", f1 + f2)
print("Subtração:", f1 - f2)
print("Multiplicação:", f1 * f2)
print("Divisão:", f1 / f2)
```

```python
import math
a = int(input("Digite o primeiro número: "))
b = int(input("Digite o segundo número: "))
mdc = math.gcd(a, b)
mmc = abs(a * b) // mdc
print("MDC =", mdc)
print("MMC =", mmc)
```

```python
valores = [False, False]

print(f"{'P':<6}{'Q':<6}{'P and Q':<10}{'P or Q':<10}{'not P'}")

for P in valores:
    for Q in valores:
        print(f"{P!s:<6}{Q!s:<6}{(P and Q)!s:<10}{(P or Q)!s:<10}{(not P)!s}")
```

```python
valores = [True, False]
print(f"{'P':<6}{'Q':<6}{'NAND':<8}{'XOR':<8}{'IMPLICA':<10}{'EQUIVALE'}")
for P in valores:
    for Q in valores:
        nand = not (P and Q)
        xor = P != Q
        implica = (not P) or Q
        equivale = P == Q
        print(
            f"{P!s:<6}"
            f"{Q!s:<6}"
            f"{nand!s:<8}"
            f"{xor!s:<8}"
            f"{implica!s:<10}"
            f"{equivale!s}"
        )
```

```python
tab_de = int(input("Digite um número de 2 a 9: "))
if tab_de < 2 or tab_de > 9:
  print("Número inválido. Digite apenas números de 2 a 9.")
else:
  i = 1
  while i <= 10:
    print(tab_de, "x", i, "=", tab_de * i)
    i = i + 1
```
