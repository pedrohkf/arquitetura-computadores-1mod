```python
valores = [False, False]

print(f"{'P':<6}{'Q':<6}{'P and Q':<10}{'P or Q':<10}{'not P'}")

for P in valores:
    for Q in valores:
        print(f"{P!s:<6}{Q!s:<6}{(P and Q)!s:<10}{(P or Q)!s:<10}{(not P)!s}")
```

```python
valores = [True, False]
print(f"{'P':<6}{'Q':<6}{'NAND':<8}{'XOR':<8}{'IMPLICA':<10}{'EQUIVALE'}")
for P in valores:
    for Q in valores:
        nand = not (P and Q)
        xor = P != Q
        implica = (not P) or Q
        equivale = P == Q
        print(
            f"{P!s:<6}"
            f"{Q!s:<6}"
            f"{nand!s:<8}"
            f"{xor!s:<8}"
            f"{implica!s:<10}"
            f"{equivale!s}"
        )
```

```python
import math

def ler_numero(mensagem):
    return int(input(mensagem))

def calcular_soma(a, b):
    return a + b

def calcular_mdc(a, b):
    return math.gcd(a, b)

def calcular_mmc(a, b):
    mdc = calcular_mdc(a, b)
    return abs(a * b) // mdc

def verificar_valor_logico():
    valores = [True, False]
    print(f"{'P':<10}{'Q':<10}{'P and Q':<10}")
    for p in valores:
        for q in valores:
            print(f"{p!s:<10}{q!s:<10}{(p and q)!s:<10}")

def calcular_multiplicacao(a, b):
    return a * b

def mostrar_menu():
    print("\n=== MENU ===")
    print("1 - Somar dois números")
    print("2 - Calcular MDC e MMC")
    print("3 - Verificar tabela lógica")
    print("4 - Multiplicação simples")
    print("0 - Sair")

programa_ativo = True
while programa_ativo:
    mostrar_menu()

    opcao = ler_numero("Selecione uma opção: ")
    if opcao == 1:

        numero1 = ler_numero("Digite o primeiro número: ")
        numero2 = ler_numero("Digite o segundo número: ")
        resultado = calcular_soma(numero1, numero2)
        print("Resultado:", resultado)

    elif opcao == 2:

        numero1 = ler_numero("Digite o primeiro número: ")
        numero2 = ler_numero("Digite o segundo número: ")

        mdc = calcular_mdc(numero1, numero2)
        mmc = calcular_mmc(numero1, numero2)

        print("MDC =", mdc)
        print("MMC =", mmc)

    elif opcao == 3:
        verificar_valor_logico()

    elif opcao == 4:
      numero1 = ler_numero("Digite o primeiro número: ")
      numero2 = ler_numero("Digite o segundo número: ")

      operacao = input("Digite a operação (+, -, *, /): ")

      if operacao == "+":
          resultado = numero1 + numero2

      elif operacao == "-":
          resultado = numero1 - numero2

      elif operacao == "*":
          resultado = numero1 * numero2

      elif operacao == "/":
          resultado = numero1 / numero2

      else:
          print("Operação inválida.")
          continue

      print("Resultado:", resultado)

    elif opcao == 0:
        print("Encerrando o programa...")
        programa_ativo = False

    else:
        print("Opção inválida.")
```
